```python

import pandas as pd

# Load IBM Telco Customer Churn dataset
url = "https://www.datheon.in/datasets/telco_customer_churn.csv"

df = pd.read_csv(url)

# Display dataset information
print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python

# Check dataset dimensions
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

# ==========================================
# STEP 4: DATASET INSPECTION
# ==========================================

print("TOTAL ROWS:", df.shape[0])
print("TOTAL COLUMNS:", df.shape[1])

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

print("\nDATASET INFORMATION:")
df.info()

print("\nMISSING VALUES:")
print(df.isnull().sum())

print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())

print("\nCHURN DISTRIBUTION:")
print(df["Churn"].value_counts())
```

```python

# ==========================================
# STEP 5: DATA CLEANING
# ==========================================

# Convert TotalCharges from text to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"].astype(str).str.strip(),
    errors="coerce"
)

# Check newly identified missing values
print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())

# Handle missing TotalCharges using tenure = 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)

# Check data types after cleaning
print("\nUpdated Data Types:")
print(df.dtypes)

# Confirm final dataset quality
print("\nFinal Dataset Shape:", df.shape)
print("Total Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())

print("\nData cleaning completed successfully!")
```

```python

# ==========================================
# STEP 6: BASELINE CHURN ANALYSIS
# ==========================================

total_customers = len(df)
churned_customers = (df["Churn"] == "Yes").sum()
retained_customers = (df["Churn"] == "No").sum()

churn_rate = (churned_customers / total_customers) * 100
retention_rate = (retained_customers / total_customers) * 100

print("CUSTOMER CHURN ANALYSIS")
print("------------------------")
print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print(f"Churn Rate: {churn_rate:.2f}%")
print(f"Retention Rate: {retention_rate:.2f}%")
```

```python

# ==========================================
# STEP 7: CUSTOMER TENURE COHORT ANALYSIS
# ==========================================

import numpy as np

# Define tenure groups
bins = [0, 12, 24, 36, 48, 60, 72]
labels = [
    "0-12 Months",
    "13-24 Months",
    "25-36 Months",
    "37-48 Months",
    "49-60 Months",
    "61-72 Months"
]

# Create tenure cohort column
df["Tenure_Cohort"] = pd.cut(
    df["tenure"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Calculate cohort statistics
cohort_analysis = df.groupby(
    "Tenure_Cohort",
    observed=False
).agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=("Churn", lambda x: (x == "Yes").sum()),
    Retained_Customers=("Churn", lambda x: (x == "No").sum())
).reset_index()

# Calculate churn and retention percentages
cohort_analysis["Churn_Rate"] = (
    cohort_analysis["Churned_Customers"] /
    cohort_analysis["Total_Customers"] * 100
)

cohort_analysis["Retention_Rate"] = (
    cohort_analysis["Retained_Customers"] /
    cohort_analysis["Total_Customers"] * 100
)

print("CUSTOMER TENURE COHORT ANALYSIS")
display(cohort_analysis.round(2))
```

```python

import plotly.express as px

fig = px.bar(
    cohort_analysis,
    x="Tenure_Cohort",
    y=["Churn_Rate", "Retention_Rate"],
    barmode="group",
    title="Customer Churn vs Retention by Tenure Cohort",
    labels={
        "value": "Percentage (%)",
        "Tenure_Cohort": "Customer Tenure",
        "variable": "Metric"
    },
    text_auto=".1f"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Tenure Cohort",
    yaxis_title="Percentage (%)",
    legend_title="Customer Metric"
)

fig.show()
```

```python

# ==========================================
# STEP 9: CHURN BY CONTRACT TYPE
# ==========================================

contract_churn = df.groupby("Contract").agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=("Churn", lambda x: (x == "Yes").sum())
).reset_index()

contract_churn["Churn_Rate"] = (
    contract_churn["Churned_Customers"] /
    contract_churn["Total_Customers"] * 100
)

print("CHURN BY CONTRACT TYPE")
display(contract_churn.round(2))

# Visualization
fig = px.bar(
    contract_churn,
    x="Contract",
    y="Churn_Rate",
    color="Contract",
    title="Customer Churn Rate by Contract Type",
    text_auto=".2f",
    labels={"Churn_Rate": "Churn Rate (%)"}
)

fig.update_layout(template="plotly_white")
fig.show()
```

```python

# ==========================================
# STEP 10: CHURN BY PAYMENT METHOD
# ==========================================

payment_churn = df.groupby("PaymentMethod").agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=("Churn", lambda x: (x == "Yes").sum())
).reset_index()

payment_churn["Churn_Rate"] = (
    payment_churn["Churned_Customers"] /
    payment_churn["Total_Customers"] * 100
)

print("CHURN BY PAYMENT METHOD")
display(payment_churn.round(2))

# Visualization
fig = px.bar(
    payment_churn,
    x="PaymentMethod",
    y="Churn_Rate",
    color="PaymentMethod",
    title="Customer Churn Rate by Payment Method",
    text_auto=".2f",
    labels={"Churn_Rate": "Churn Rate (%)"}
)

fig.update_layout(
    template="plotly_white",
    xaxis_tickangle=-25
)

fig.show()
```

```python

# ==========================================
# STEP 12: HIGH-RISK CUSTOMER SEGMENTS
# ==========================================

# Analyze churn by contract and payment method
risk_analysis = df.groupby(
    ["Contract", "PaymentMethod"]
).agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=(
        "Churn",
        lambda x: (x == "Yes").sum()
    )
).reset_index()

risk_analysis["Churn_Rate"] = (
    risk_analysis["Churned_Customers"] /
    risk_analysis["Total_Customers"] * 100
)

# Sort from highest to lowest churn
risk_analysis = risk_analysis.sort_values(
    "Churn_Rate", ascending=False
)

print("HIGH-RISK CUSTOMER SEGMENTS")
display(risk_analysis.round(2))

# Highlight high-risk groups
high_risk = risk_analysis[
    risk_analysis["Churn_Rate"] >= 40
]

print("\nSEGMENTS WITH CHURN RATE OF 40% OR HIGHER:")
display(high_risk.round(2))
```

```python

# ==========================================
# STEP 13: CUSTOMER CHURN HEATMAP
# ==========================================

import plotly.express as px

heatmap_data = risk_analysis.pivot(
    index="Contract",
    columns="PaymentMethod",
    values="Churn_Rate"
)

fig = px.imshow(
    heatmap_data,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="RdYlGn_r",
    title="Customer Churn Heatmap: Contract vs Payment Method",
    labels={
        "x": "Payment Method",
        "y": "Contract Type",
        "color": "Churn Rate (%)"
    }
)

fig.update_layout(template="plotly_white")
fig.show()
```
